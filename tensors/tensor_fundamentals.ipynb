## Tensor Properties

```python
A = torch.rand(3, 3, 2,
               dtype=torch.float16,
               device=None,
               requires_grad=False)

# The number of dimensions 

A.ndim

# The number along each dimension out to in

A.shape

# The datatype stored inside the tensor

A.dtype

# The device the tensor is stored on 

A.device
```

## Tensor Operations


```python

A = torch.rand(3, 3)
B = torch.rand(3, 3)

# Elementwise & Scalar Ops

## torch.add() torch.mul() torch.sub() for scalar ops

## Addiition

C = A + B
A = A + 10

## Subtraction

C = A - B
A = A - 10

## Multiplication (Haramard)

C = A * B
A = A * 10

## Division (Haramard?)

C = A / B
A = A / 10

## Matrix Multiplication torch.matmul(a, b) torch.mm()

C = A @ B

# Transpose

A.T

A.transpose(A, dim0, dim1) # Swaps dim0 and dim1 in the tensor
```
